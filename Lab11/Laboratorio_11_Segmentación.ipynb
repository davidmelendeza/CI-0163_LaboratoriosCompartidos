```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Configuración para evitar el warning de multiprocessing con KMeans
import os
os.environ['LOKY_MAX_CPU_COUNT'] = '4'


# Cargar datasets como archivos .csv
df_lunas = pd.read_csv('lunas.csv')
df_clusters = pd.read_csv('clusters_dispersos.csv')
df_circulos = pd.read_csv('circulos_concentricos.csv')

#Visualizacion de los datos
print("Lunas DataFrame:")
print(df_lunas.head())
print("\nClusters Dispersos DataFrame:")
print(df_clusters.head())
print("\nCírculos Concéntricos DataFrame:")
print(df_circulos.head())
```

## Dataset de Lunas

### DBScan

```python
# Escalamiento
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_lunas)

# Aplicar DBScan
dbscan = DBSCAN(eps=0.3, min_samples=5)
clusters = dbscan.fit_predict(X_scaled)

# Agregar etiquetas al DataFrame
df_lunas['Cluster'] = clusters

# Visualizar los resultados
plt.figure(figsize=(8, 6))
plt.scatter(df_lunas['Feature1'], df_lunas['Feature2'], c=df_lunas['Cluster'], cmap='viridis', s=50)
plt.title('Clusters generados por DBSCAN')
plt.xlabel('Feature1')
plt.ylabel('Feature2')
plt.grid(True)
plt.show()

# Ver cuántos clusters encontró (incluyendo -1 para ruido)
print(df_lunas['Cluster'].value_counts().sort_index())
```

### KMeans

```python
# Escalamiento
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_lunas)

# Aplicar KMeans (elige k clusters)
k = 2
kmeans = KMeans(n_clusters=k, random_state=0)
clusters = kmeans.fit_predict(X_scaled)

# Agregar etiquetas al DataFrame
df_lunas['Cluster'] = clusters

# Visualizar los resultados
plt.figure(figsize=(8, 6))
plt.scatter(df_lunas['Feature1'], df_lunas['Feature2'], c=df_lunas['Cluster'], cmap='viridis', s=50)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], c='red', s=200, marker='X', label='Centroides')
plt.title('Clusters generados por KMeans')
plt.xlabel('Feature1')
plt.ylabel('Feature2')
plt.grid(True)
plt.legend()
plt.show()

# Ver cuántos puntos hay en cada cluster
print(df_lunas['Cluster'].value_counts().sort_index())
```
